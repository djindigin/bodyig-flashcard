```python
# 安装处理音频及打包所需的库
!pip install -q pydub
!apt-get install -qq ffmpeg

import os
import zipfile
from pydub import AudioSegment
from pydub.silence import split_on_silence
print("环境准备就绪！")
```

```python
from google.colab import files

# 1. 上传音频文件
uploaded = files.upload()
input_filename = list(uploaded.keys())[0]
file_ext = input_filename.split('.')[-1].lower()

# 2. 加载音频
audio = AudioSegment.from_file(input_filename, format=file_ext)

# 3. 按照静音自动切分
# 可根据需要微调 min_silence_len（静音最短长度ms）和 silence_thresh（阈值）
chunks = split_on_silence(
    audio,
    min_silence_len=400,
    silence_thresh=audio.dBFS - 16,
    keep_silence=100
)

# 4. 导出并按三位数字编号命名 (001.m4a, 002.m4a...)
output_dir = "split_output"
os.makedirs(output_dir, exist_ok=True)
saved_files = []

for i, chunk in enumerate(chunks):
    out_name = f"{i+1:03d}.{file_ext}"
    out_path = os.path.join(output_dir, out_name)
    # Fix: Specify 'mp4' format and 'aac' codec for .m4a output
    chunk.export(out_path, format='mp4', codec='aac')
    saved_files.append(out_path)

print(f"切分完成！共切出 {len(chunks)} 段音频。")

# 5. 打包成 ZIP
zip_filename = "split_audio_package.zip"
with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for file in saved_files:
        zipf.write(file, arcname=os.path.basename(file))

print(f"压缩包已生成: {zip_filename}")

# 6. 自动触发浏览器下载
files.download(zip_filename)
```
